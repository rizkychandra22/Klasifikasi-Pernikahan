```python
import os
import glob
import pandas as pd
import numpy as np

raw_dir = '..' if os.path.exists(os.path.join('..', 'Data-Excel')) else '.'
data_excel = os.path.join(raw_dir, 'Data-Excel') if os.path.exists(os.path.join(raw_dir, 'Data-Excel')) else raw_dir
dataset_dir = 'dataset' if os.path.exists('dataset') else os.path.join('..', 'dataset')
os.makedirs(dataset_dir, exist_ok=True)
```

```python
def extract_clean(file_paths, year):
    dfs = []
    for f in sorted(file_paths):
        df = pd.read_excel(f)
        sub = pd.DataFrame()
        
        sub['Nama Suami'] = df['Nama Suami'].astype(str).str.strip()
        sub['Nama Istri'] = df['Nama Istri'].astype(str).str.strip()
        
        col_us = 'Umur Suami' if 'Umur Suami' in df.columns else 'Usia Suami'
        col_ui = 'Umur Istri' if 'Umur Istri' in df.columns else 'Usia Istri'
        sub['Usia Suami'] = df[col_us].astype(str).str.extract(r'(\d+)')[0].astype(int)
        sub['Usia Istri'] = df[col_ui].astype(str).str.extract(r'(\d+)')[0].astype(int)
        
        sub['Pendidikan Suami'] = df['Pendidikan Suami'].astype(str).str.strip().replace({'nan': np.nan, 'None': np.nan})
        sub['Pendidikan Istri'] = df['Pendidikan Istri'].astype(str).str.strip().replace({'nan': np.nan, 'None': np.nan})
        
        sub['Pekerjaan Suami'] = df['Pekerjaan Suami'].astype(str).str.strip()
        sub['Pekerjaan Istri'] = df['Pekerjaan Istri'].astype(str).str.strip()
        sub['Tahun'] = year
        dfs.append(sub)
        
    return pd.concat(dfs, ignore_index=True)
```

```python
tasks = [
    ('Nyalindung 2021', 'nyalindung_21.csv', 2021),
    ('Nyalindung 2022', 'nyalindung_22.csv', 2022),
    ('Nyalindung 2023', 'nyalindung_23.csv', 2023),
    ('Nyalindung 2024', 'nyalindung_24.csv', 2024),
    ('Nyalindung 2025', 'nyalindung_25.csv', 2025),
    ('Nyalindung 2026', 'nyalindung_26.csv', 2026)
]

for folder, outfile, year in tasks:
    files = sorted(glob.glob(os.path.join(data_excel, folder, '*.xlsx')))
    if files:
        df_clean = extract_clean(files, year)
        df_clean.to_csv(os.path.join(dataset_dir, outfile), index=False, encoding='utf-8')
        print(f'{outfile}: {len(df_clean)} rows')
```

```python
csv_files = [
    'nyalindung_21.csv', 'nyalindung_22.csv', 'nyalindung_23.csv',
    'nyalindung_24.csv', 'nyalindung_25.csv', 'nyalindung_26.csv'
]

dfs = [pd.read_csv(os.path.join(dataset_dir, f)) for f in csv_files if os.path.exists(os.path.join(dataset_dir, f))]
index_data = pd.concat(dfs, ignore_index=True)
index_data.to_csv(os.path.join(dataset_dir, 'index_data.csv'), index=False, encoding='utf-8')
print(f'index_data.csv: {len(index_data)} rows')
```
